```python
import json
import math
import numpy as np
```

```python
with open("bootstrapResults.json", 'r') as file:
    # Load the JSON data from the file object
    stats = json.load(file)
```

```python
YOLOv10Boot = stats["YOLOv10nData"]
YOLOv11Boot = stats["YOLOv11nData"]
YOLOv12Boot = stats["YOLOv12nData"]
```

```python
precisionValuev10 = []
recallValuev10 = []
F1ScoreValuev10 = []
ap5095Valuev10 = []
ap50Valuev10 = []
fitnessValuev10 = []

for metrics in YOLOv10Boot:
    precisionValuev10.append(metrics[0])
    recallValuev10.append(metrics[1])
    F1ScoreValuev10.append(2 * ((metrics[0]*metrics[1])/(metrics[0]+metrics[1])))
    ap5095Valuev10.append(metrics[2])
    ap50Valuev10.append(metrics[3])
    fitnessValuev10.append((metrics[2]*0.9) + (metrics[3]*0.1))
```

```python
print("Number of values: ", len(precisionValuev10))
print(f'Mean precision for YOLOv10:  {np.mean(precisionValuev10)} ({precisionValuev10[24]}, {precisionValuev10[974]})')
print(f'Mean recall for YOLOv10:     {np.mean(recallValuev10)} ({recallValuev10[24]}, {recallValuev10[974]})')
print(f'Mean F1 for YOLOv10:         {np.mean(F1ScoreValuev10)} ({F1ScoreValuev10[24]}, {F1ScoreValuev10[974]})')
print(f'Mean AP5095 for YOLOv10:     {np.mean(ap5095Valuev10)} ({ap5095Valuev10[24]}, {ap5095Valuev10[974]})')
print(f'Mean AP50 for YOLOv10:       {np.mean(ap50Valuev10)} ({ap50Valuev10[24]}, {ap50Valuev10[974]})')
print(f'Mean fitness for YOLOv10:    {np.mean(fitnessValuev10)} ({fitnessValuev10[24]}, {fitnessValuev10[974]})')
```
